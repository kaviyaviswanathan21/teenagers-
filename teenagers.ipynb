# Does the Specific Month a Student was Born in Impact their Favorite Season?

To answer this, I will be conducting a **census study** because I will be using data from the entire population of teenagers around America. I will be checking if there is some type of **association** between a student's favorite season and the month they were born. I predict that a student's favorite season would be whatever season occurs during that specific month they were born. 

```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt

teenagers = pd.read_csv("CensusAtSchool.csv")
teenagers.head()
```

## Overview:

```python
teenagers.info()
```

## Visuals:

### Side-by-Side Graphs

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

season_colors = {
    "Spring": "#66BB6A",   # green
    "Summer": "#FFD54F",   # yellow
    "Fall": "#FF8A65",   # orange
    "Winter": "#64B5F6"    # blue
}

sns.countplot(
    data=teenagers,
    x="Birth_month",
    hue="Favorite_Season",
    order=month_order,
    palette=season_colors
)

plt.title("Favorite Season by Birth Month")
plt.xlabel("Birth Month")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.legend(title="Favorite Season")
plt.tight_layout()
plt.show()
```

```python
ax = sns.countplot(
data=teenagers,
x="Birth_month",
hue="Favorite_Season",
order=month_order,
palette=season_colors
)
 
for container in ax.containers:
    ax.bar_label(container)
 
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Table:

```python
table = pd.crosstab(
    teenagers['Birth_month'],
    teenagers['Favorite_Season'],
    margins=True,
    margins_name='Total'
)

print(table)
```
